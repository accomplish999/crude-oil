# 01 Seasonality

Do WTI's calendar-month mean returns from before 2020 show up again after it?

Holdout starts 2020-01-01. Verdict: held.

This study reads stored series. It does not add a derived column.

```python
import json
from pathlib import Path
studies = json.loads(Path('data/studies.json').read_text())
study = next(item for item in studies['studies'] if item['slug'] == 'seasonality')
print(study['verdict'])
print(study['verdict_line'])
print(study['table']['caption'])
for row in study['table']['rows']:
    print(' | '.join(row))
```

## Method

Train prints end 31 Dec 2019. Holdout prints start 1 Jan 2020. Each calendar month gets a mean daily log return in each window.

## Result

Correlation of the 12 train month-means with the 12 holdout month-means is 0.309. Signs match in 5 of 12 months. The correlation is above zero, which was the bar. 5 matching signs out of 12 is a thin map.
